```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class BMIState(TypedDict):
    height_m : int
    weight_kg : int
    bmi : float
```

```python
graph = StateGraph(BMIState)
```

```python
def bmicalculator(state:BMIState)->BMIState:
    bmi = state["height_m"] / state["weight_kg"]
    return {"bmi":bmi}
```

```python
graph.add_node("bmicalculatornode",bmicalculator)
graph.add_edge(START,'bmicalculatornode')
graph.add_edge('bmicalculatornode',END)
```

```python
workflow = graph.compile()
workflow
```

```python
initial_state = {
    "height_m" : 14,
    "weight_kg" : 2
}
final_state = workflow.invoke(initial_state)
```
