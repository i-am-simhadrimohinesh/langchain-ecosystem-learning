```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.prompts import MessagesPlaceholder
from langchain_core.messages import AIMessage,SystemMessage,HumanMessage
```

```python
chat_history = [
        HumanMessage('what is AI fullform'),
        AIMessage('AI stands for Artificial Intelligence')
    ]
```

```python
PromptTemplate1 = ChatPromptTemplate([
    # ("system", "You are a helpful assistant."),
    SystemMessage(content="You are a helpful assistant."),
    MessagesPlaceholder("history"),
    HumanMessage(content="{userInput}")
    # ("human", "{userInput}")
])
```

```python
prompt = PromptTemplate1.invoke({
    'userInput':'Tell me more about that',
    'history' : chat_history
})
```
