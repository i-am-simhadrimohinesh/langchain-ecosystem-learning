```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = GoogleGenerativeAIEmbeddings(model='gemini-embedding-2')
```

```python
doc = 'LangChain, LangGraph and LangSmith are the frameworks in the langchain ecosystem'
```

```python
res1 = model.embed_query(doc)
```

```python
docs = [
        "The cat sat on the mat.",
        "A feline is sitting on a rug.",
        "Today's weather is sunny."
    ]
```

```python
res2 = model.embed_documents(docs)
```
